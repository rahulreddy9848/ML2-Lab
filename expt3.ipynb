```python

import pandas as pd
from google.colab import files


print("Upload your Kaggle Play Tennis CSV file:")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

data = pd.read_csv(filename)

print("\nDataset loaded successfully!")
print("Shape:", data.shape)

print("\nFirst 5 rows:")
print(data.head())



if "day" in data.columns:
    data = data.drop("day", axis=1)

target = "play"

data = data.dropna()

print("\nDataset after preprocessing:")
print(data.head())

print("\nNumber of rows:", len(data))



attributes = list(data.columns)

attributes.remove(target)

hypothesis = ["∅"] * len(attributes)

print("\nAttributes:")
print(attributes)

print("\nInitial Hypothesis:")
print(hypothesis)



for index, row in data.iterrows():


    if str(row[target]).strip().lower() == "yes":

        print("\n--------------------------------")
        print("Positive Example:", index + 1)
        print("--------------------------------")

        for i in range(len(attributes)):


            if hypothesis[i] == "∅":
                hypothesis[i] = row[attributes[i]]


            elif hypothesis[i] != row[attributes[i]]:
                hypothesis[i] = "?"

        print("Current Hypothesis:")
        print(hypothesis)



print("\n============================================")
print("FINAL HYPOTHESIS")
print("============================================")

for attribute, value in zip(attributes, hypothesis):
    print(attribute, "=", value)

print("\nMost Specific Hypothesis:")
print(hypothesis)
```

```python
import pandas as pd


url = "https://archive.ics.uci.edu/ml/machine-learning-databases/mushroom/agaricus-lepiota.data"

print("Downloading Mushroom dataset...")

data = pd.read_csv(url, header=None)

print("\nDataset loaded successfully!")
print("Shape:", data.shape)



columns = [
    "class",
    "cap_shape",
    "cap_surface",
    "cap_color",
    "bruises",
    "odor",
    "gill_attachment",
    "gill_spacing",
    "gill_size",
    "gill_color",
    "stalk_shape",
    "stalk_root",
    "stalk_surface_above_ring",
    "stalk_surface_below_ring",
    "stalk_color_above_ring",
    "stalk_color_below_ring",
    "veil_type",
    "veil_color",
    "ring_number",
    "ring_type",
    "spore_print_color",
    "population",
    "habitat"
]

data.columns = columns



print("\nFirst 5 rows:")
print(data.head())

print("\nDataset shape:")
print(data.shape)

print("\nColumns:")
print(data.columns.tolist())




print("\nMissing values before preprocessing:")
print((data == "?").sum().sum())

data = data.replace("?", pd.NA)

data = data.dropna()

print("\nDataset after removing missing values:")
print(data.shape)



target = "class"

attributes = list(data.columns)

attributes.remove(target)

hypothesis = ["∅"] * len(attributes)

print("\n============================================")
print("FIND-S ALGORITHM")
print("============================================")

print("\nNumber of attributes:", len(attributes))

print("\nAttributes:")
print(attributes)

print("\nInitial Hypothesis:")
print(hypothesis)



positive_count = 0

for index, row in data.iterrows():





    if str(row[target]).strip().lower() == "e":

        positive_count += 1

        print("\n--------------------------------")
        print("Positive Example:", index + 1)
        print("--------------------------------")

        for i in range(len(attributes)):


            if hypothesis[i] == "∅":

                hypothesis[i] = row[attributes[i]]


            elif hypothesis[i] != row[attributes[i]]:

                hypothesis[i] = "?"

        print("Current Hypothesis:")
        print(hypothesis)



print("\n============================================")
print("FINAL RESULT")
print("============================================")

print("\nNumber of positive examples:", positive_count)

print("\nFinal Hypothesis:")

for attribute, value in zip(attributes, hypothesis):

    print(attribute, "=", value)

print("\nMost Specific Hypothesis:")
print(hypothesis)
```
